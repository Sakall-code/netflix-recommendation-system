```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
from surprise import SVD, Dataset, accuracy
from surprise.model_selection import KFold

# ==========================================
# PART 1: DATA GENERATION (50-Fold CV)
# ==========================================
print("Loading data and starting 50-Fold Cross-Validation...")
data = Dataset.load_builtin('ml-100k')
algo = SVD(random_state=42)
kf = KFold(n_splits=50, random_state=42, shuffle=True)

rmse_scores = []
mae_scores = []
fold_indices = []

# Training Loop
for i, (trainset, testset) in enumerate(kf.split(data), 1):
    algo.fit(trainset)
    predictions = algo.test(testset)

    rmse_scores.append(accuracy.rmse(predictions, verbose=False))
    mae_scores.append(accuracy.mae(predictions, verbose=False))
    fold_indices.append(i)

# Create DataFrame
df_results = pd.DataFrame({
    'Fold': fold_indices,
    'RMSE': rmse_scores,
    'MAE': mae_scores
})

# Calculate Statistics
mean_rmse = np.mean(rmse_scores)
mean_mae = np.mean(mae_scores)

print(f"\nProcessing Complete. Mean RMSE: {mean_rmse:.4f} | Mean MAE: {mean_mae:.4f}")

# ==========================================
# PART 2: VISUALIZATION GENERATION
# ==========================================
sns.set_style("whitegrid")
plt.rcParams.update({'font.size': 12, 'font.family': 'sans-serif'})

# --- FIG X: Fold-wise RMSE Performance (Line Plot) ---
plt.figure(figsize=(12, 5))
plt.plot(df_results['Fold'], df_results['RMSE'], marker='o', color='#1f77b4', linewidth=1.5, markersize=4)
plt.axhline(mean_rmse, color='red', linestyle='--', label=f'Mean RMSE ({mean_rmse:.3f})')
plt.title('Fig. X. Fold-wise RMSE Performance (50 Folds)', fontsize=14, pad=15)
plt.xlabel('Fold Number', fontsize=12)
plt.ylabel('RMSE', fontsize=12)
plt.legend()
plt.xlim(0, 51)
plt.tight_layout()
plt.savefig('Fig_X_Fold_RMSE.pdf', format='pdf', dpi=300)
plt.show()

# --- FIG Y: Fold-wise MAE Performance (Line Plot) ---
plt.figure(figsize=(12, 5))
plt.plot(df_results['Fold'], df_results['MAE'], marker='s', color='#2ca02c', linewidth=1.5, markersize=4)
plt.axhline(mean_mae, color='red', linestyle='--', label=f'Mean MAE ({mean_mae:.3f})')
plt.title('Fig. Y. Fold-wise MAE Performance (50 Folds)', fontsize=14, pad=15)
plt.xlabel('Fold Number', fontsize=12)
plt.ylabel('MAE', fontsize=12)
plt.legend()
plt.xlim(0, 51)
plt.tight_layout()
plt.savefig('Fig_Y_Fold_MAE.pdf', format='pdf', dpi=300)
plt.show()

# --- FIG Z: Mean Error Comparison (Bar Chart) ---
plt.figure(figsize=(7, 5))
metrics = ['RMSE', 'MAE']
means = [mean_rmse, mean_mae]
stds = [np.std(rmse_scores), np.std(mae_scores)]
colors = ['#1f77b4', '#2ca02c']

bars = plt.bar(metrics, means, yerr=stds, capsize=10, color=colors, alpha=0.8, width=0.5)
plt.title('Fig. Z. Mean Error Comparison (RMSE vs MAE)', fontsize=14, pad=15)
plt.ylabel('Error Value', fontsize=12)
plt.ylim(0, 1.1)

# Add value labels on bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 0.02,
             f'{height:.3f}', ha='center', va='bottom', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.savefig('Fig_Z_Mean_Comparison.pdf', format='pdf', dpi=300)
plt.show()

# --- FIG A: Statistical Robustness (Violin & Box Plot) ---
# Reshape data for seaborn
df_long = df_results.melt(id_vars='Fold', value_vars=['RMSE', 'MAE'], var_name='Metric', value_name='Error')

plt.figure(figsize=(8, 6))
# Violin plot for distribution shape
sns.violinplot(x='Metric', y='Error', data=df_long, inner=None, palette="pastel")
# Box plot for statistical quartiles
sns.boxplot(x='Metric', y='Error', data=df_long, width=0.2, color='white', boxprops={'zorder': 2})

plt.title('Statistical Robustness Analysis (Distribution over 50 Folds)', fontsize=14, pad=15)
plt.ylabel('Error Distribution', fontsize=12)
plt.tight_layout()
plt.savefig('Fig_A_Statistical_Robustness.pdf', format='pdf', dpi=300)
plt.show()

# --- FIG B: Overlay Comparison (RMSE vs MAE) ---
plt.figure(figsize=(12, 6))
plt.plot(df_results['Fold'], df_results['RMSE'], label='RMSE', color='#1f77b4', alpha=0.8)
plt.plot(df_results['Fold'], df_results['MAE'], label='MAE', color='#2ca02c', alpha=0.8)
plt.fill_between(df_results['Fold'], df_results['RMSE'], df_results['MAE'], color='gray', alpha=0.1)

plt.title('Comparative Error Analysis: RMSE vs MAE across 50 Folds', fontsize=14, pad=15)
plt.xlabel('Fold Number', fontsize=12)
plt.ylabel('Error Magnitude', fontsize=12)
plt.legend(loc='upper right', frameon=True)
plt.xlim(0, 51)
plt.tight_layout()
plt.savefig('Fig_B_Overlay_Comparison.pdf', format='pdf', dpi=300)
plt.show()

print("All figures saved as PDF successfully.")
```
